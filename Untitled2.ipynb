## 1. Formulación del Problema y Marco Teórico

El **método de bisección** es un algoritmo de búsqueda de raíces que divide repetidamente un intervalo a la mitad y selecciona un subintervalo en el que debe residir una raíz para su procesamiento posterior.

Dada una función continua $f[a, b]$ donde $f(a)$ y $f(b)$ tienen signos opuestos (es decir, $f(a) \cdot f(b) < 0$), el Teorema del Valor Intermedio garantiza la existencia de al menos una solución para la ecuación $f(x) = 0$ en el intervalo $[a, b]$ [1, 2].

### Definición del Ejemplo 1
Queremos demostrar que $f(x) = x^3 + 4x^2 - 10 = 0$ tiene una raíz en el intervalo $[1, 2]$ y utilizar el método de bisección para determinar una aproximación con una tolerancia o precisión deseada[cite: 2].

```python
#Importación de librerías necesarias para gráficos y manejo numérico
import numpy as np
import matplotlib.pyplot as plt

def biseccion(f, a, b, tol, N0):
    """
    Implementación del Algoritmo 2.1 (Método de Bisección) de Burden & Faires.

    Parámetros:
    -----------
    f : function
        La función f(x) a evaluar.
    a : float
        Extremo izquierdo del intervalo inicial.
    b : float
        Extremo derecho del intervalo inicial.
    tol : float
        Tolerancia o criterio de paro para el error o tamaño del intervalo.
    N0 : int
        Número máximo de iteraciones permitidas.

    Retorna:
    --------
    p : float
        Aproximación de la raíz.
    historial : list
        Lista de tuplas que contiene los datos de cada iteración para análisis.
    """

    #Paso 1: Inicialización de variables
    i = 1
    FA = f(a)

    # Historial para almacenar los resultados tabulados por iteración
    historial = []

    #Paso 2: Ciclo principal condicionado al número máximo de iteraciones N0
    while i <= N0:
        #Paso 3: Cálculo del punto medio p (usando la fórmula estable para evitar desbordamientos)
        p = a + (b - a) / 2.0
        FP = f(p)

        #Guardar el registro de la iteración actual
        historial.append((i, a, b, p, FP))

        #Paso 4: Criterio de paro (si se encuentra la raíz exacta o el intervalo es menor que la tolerancia)
        if FP == 0.0 or (b - a) / 2.0 < tol:
            print(f"Procedimiento completado exitosamente en la iteración {i}.")
            return p, historial

        #Paso 5: Incremento del contador de iteraciones
        i += 1

        #Paso 6: Determinación del nuevo subintervalo según los signos
        #Se utiliza el producto FA * FP para evaluar el cambio de signo
        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p  # Si FA * FP < 0, el cambio ocurre entre a y p; b se desplaza a p (FA no cambia)

    #Paso 7: Si se agotan las iteraciones sin cumplir el criterio
    raise ValueError(f"El método fracasó después de {N0} iteraciones (procedimiento no exitoso).")

#==========================================
#Ejecución del Ejemplo 1 de Burden (p. 38)
#==========================================

#Definición de la función del Ejemplo 1: f(x) = x^3 + 4x^2 - 10
def f_ejemplo1(x):
    return x**3 + 4*x**2 - 10

#Parámetros iniciales indicados en el libro
a_ini = 1.0
b_ini = 2.0
tolerancia = 1e-4  #Precisión dentro de 10^(-4)
max_iter = 100

#Verificación inicial del Teorema del Valor Intermedio
if f_ejemplo1(a_ini) * f_ejemplo1(b_ini) < 0:
    raiz_aprox, resultados = biseccion(f_ejemplo1, a_ini, b_ini, tolerancia, max_iter)
    print(f"\nAproximación final de la raíz (p): {raiz_aprox:.7f}")
else:
    print("El intervalo no cumple con las condiciones de cambio de signo.")
```
