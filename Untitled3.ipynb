##1. Formulación del Problema y Marco Teórico

El método de punto fijo busca resolver ecuaciones de la forma $f(x) = 0$ transformándolas algebraicamente en un problema equivalente de punto fijo:
$$p = g(p)$$

Para aproximar el punto fijo, partimos de una aproximación inicial $p_0$ y generamos la sucesión recurrente:
$$p_n = g(p_{n-1}) \quad \text{para } n \ge 1$$

###Aplicación al Ejemplo 1
Dada la ecuación original $x^3 + 4x^2 - 10 = 0$, podemos despejar y construir distintas funciones de iteración $g(x)$. En este ejercicio utilizaremos la función de la opción (e) del análisis del libro:
$$g(x) = x - \frac{x^3 + 4x^2 - 10}{3x^2 + 8x}$$
la cual converge de manera rápida y eficiente partiendo de la aproximación inicial $p_0 = 1.5$.

```python
#Importación de librerías para visualización y soporte numérico
import numpy as np
import matplotlib.pyplot as plt

def punto_fijo(g, p0, tol, N0):
    """
    Implementación del Algoritmo 2.2 (Iteración de Punto Fijo) de Burden & Faires.

    Parámetros:
    -----------
    g : function
        La función de iteración g(x).
    p0 : float
        Aproximación inicial.
    tol : float
        Tolerancia para el criterio de paro (|p - p0| < tol).
    N0 : int
        Número máximo de iteraciones permitidas.

    Retorna:
    --------
    p : float
        La aproximación final del punto fijo.
    historial : list
        Lista con los registros de cada iteración para análisis y tabulación.
    """

    #Paso 1: Inicialización de la variable de conteo de iteraciones
    i = 1

    #Historial para almacenar el seguimiento de la sucesión
    historial = []

    #Paso 2: Ciclo principal controlado por el límite de iteraciones N0
    while i <= N0:
        #Paso 3: Calcular la nueva aproximación p = g(p0)
        p = g(p0)

        #Guardar los datos de la iteración actual en el historial
        historial.append((i, p0, p, abs(p - p0)))

        #Paso 4: Criterio de paro basado en la distancia absoluta entre aproximaciones sucesivas
        if abs(p - p0) < tol:
            print(f"Procedimiento completado exitosamente en la iteración {i}.")
            return p, historial

        #Paso 5: Incrementar el contador de iteraciones
        i += 1

        #Paso 6: Actualizar p0 con el nuevo valor de p para la siguiente iteración
        p0 = p

    #Paso 7: Si se alcanza el límite sin converger
    raise ValueError(f"El método fracasó después de {N0} iteraciones (procedimiento no exitoso).")

#==========================================
#Configuración y Ejecución del Problema
#==========================================

#Definición de la función de iteración g(x) [Opción (e) de Burden, p. 46]
def g_func(x):
    return x - (x**3 + 4*x**2 - 10) / (3*x**2 + 8*x)

#Parámetros iniciales recomendados en el libro
p_inicial = 1.5
tolerancia = 1e-5  # Alta precisión
max_iter = 50

#Ejecución del algoritmo de punto fijo
raiz_aprox, resultados = punto_fijo(g_func, p_inicial, tolerancia, max_iter)
print(f"\nAproximación final del punto fijo (p): {raiz_aprox:.13f}")
```

##2. Resultados Obtenidos y Sucesión de Aproximaciones

A continuación se enlistan los valores generados en cada iteración por el Algoritmo 2.2, mostrando la rápida convergencia hacia la raíz real:

```python
#Impresión limpia en formato tabular de la evolución de p_n
print(f"{'n':<4}|{'p_(n-1)':<15}|{'p_n':<15}|{'|p_n - p_(n-1)|':<18}")
print("-" * 58)
for row in resultados:
    n, p_ant, p_act, error = row
    print(f"{n:<4}|{p_ant:<15.8f}|{p_act:<15.8f}|{error:<18.2e}")
```

##3. Visualización Gráfica

La gráfica muestra la intersección entre la función y la línea identidad $y = x$, ilustrando el comportamiento iterativo del método de punto fijo hacia la solución.

```python
#Configuración del gráfico de convergencia
plt.figure(figsize=(9, 5))
x_vals = np.linspace(1.2, 1.6, 400)
y_vals = g_func(x_vals)

#Curva de g(x) y línea identidad y = x
plt.plot(x_vals, y_vals, label='$g(x)$', color='darkorange', linewidth=2)
plt.plot(x_vals, x_vals, label='Línea $y = x$', color='gray', linestyle='--', linewidth=1.5)

#Marcar el punto final de convergencia
plt.scatter([raiz_aprox], [raiz_aprox], color='purple', zorder=5, label=f'Punto Fijo: {raiz_aprox:.7f}')

#Estética de la gráfica
plt.title('Método de Iteración de Punto Fijo', fontsize=12, fontweight='bold')
plt.xlabel('x', fontsize=11)
plt.ylabel('g(x)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10)
plt.show()
```
